```python
print("=== SISTEM PENCATATAN HASIL PANEN ===")

nama_komoditas = input("Nama komoditas: ")
jumlah = float(input("Jumlah hasil panen (kg): "))
harga = float(input("Harga per kg: "))

total = jumlah * harga

print("\n=== HASIL PENCATATAN ===")
print("Komoditas :", nama_komoditas)
print("Jumlah    :", jumlah, "kg")
print("Harga/kg  : Rp", harga)
print("Total     : Rp", total)


def hitung_diskon(total):
    if total >= 1000000:
        diskon = total * 0.10
    elif total >= 500000:
        diskon = total * 0.05
    else:
        diskon = 0

    return diskon


diskon = hitung_diskon(total)
total_akhir = total - diskon

print("Diskon    : Rp", diskon)
print("Total akhir: Rp", total_akhir)
```
